```python
# creating 3d plot using matplotlib 
# in python

# for creating a responsive plot
%matplotlib widget

# importing required libraries
from mpl_toolkits.mplot3d import Axes3D
import matplotlib.pyplot as plt
import pandas as pd

df = pd.read_csv('/home/cbyers/Projects/openmc/delta-tracking-tests/assembly/lost_assembly_particles.csv')
print(df.head(10))
print(len(df))
```

```python

# creating random dataset
xs = df['r_x'].values.tolist()
print(xs[:5])

ys = df['r_y'].values.tolist()

zs = df['r_z'].values.tolist()

# creating figure
fig = plt.figure()
ax = fig.add_subplot(projection='3d')

# creating the plot
plot_geeks = ax.scatter(xs, ys, zs, color='green')

# setting title and labels
ax.set_title("assembly Case Lost Particles")
ax.set_xlabel('x-axis')
ax.set_ylabel('y-axis')
ax.set_zlabel('z-axis')

# displaying the plot
plt.show()
```
